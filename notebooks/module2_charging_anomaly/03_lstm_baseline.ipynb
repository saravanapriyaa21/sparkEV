```python
# ============================================================
# SPARKEV MODULE 2
# MODEL 2: LSTM FOR EV CHARGING FAULT DETECTION
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import drive

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    r2_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# 2. MOUNT GOOGLE DRIVE
# ============================================================

drive.mount('/content/drive')


# ============================================================
# 3. LOAD DATASET
# ============================================================

DATA_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "processed_data_longer_than_30.xlsx"
)

df = pd.read_excel(DATA_PATH)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Dataset shape:", df.shape)
print("Number of charging sessions:",
      df["transaction_id"].nunique())

print("\nFirst 5 rows:")
display(df.head())

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 4. BASIC DATA PREPARATION
# ============================================================

lstm_df = df.copy()

lstm_df["end_time"] = pd.to_datetime(
    lstm_df["end_time"]
)

lstm_df = lstm_df.sort_values(
    ["transaction_id", "end_time"]
).reset_index(drop=True)


# ============================================================
# 5. FEATURES FOR LSTM
# ============================================================
# These represent the actual charging and battery behaviour.
# The model learns how these values change with time.

time_features = [
    "current_soc",
    "chargingv",
    "charginga",
    "out_power",
    "charging_gun_temperature1",
    "charging_gun_temperature2",
    "current_energy_meter_value",
    "total_charging_kwh",
    "total_charging_min"
]

print("\nFeatures used by LSTM:")

for feature in time_features:
    print("-", feature)


# ============================================================
# 6. KEEP ONLY REQUIRED COLUMNS
# ============================================================

required_columns = (
    ["transaction_id", "label"] +
    time_features
)

lstm_df = lstm_df[required_columns].copy()


# ============================================================
# 7. HANDLE MISSING VALUES
# ============================================================

print("\nMissing values before cleaning:")

print(
    lstm_df[time_features]
    .isnull()
    .sum()
)

lstm_df = lstm_df.dropna().reset_index(drop=True)

print("\nShape after removing missing values:",
      lstm_df.shape)


# ============================================================
# 8. CHECK SESSION LABELS
# ============================================================
# Each charging session should have one label.

session_label_check = (
    lstm_df
    .groupby("transaction_id")["label"]
    .nunique()
)

print("\nSessions with multiple labels:",
      (session_label_check > 1).sum())


# ============================================================
# 9. CREATE SESSION LABEL TABLE
# ============================================================

session_labels = (
    lstm_df
    .groupby("transaction_id")["label"]
    .first()
)

print("\nSession-level class distribution:")
print(session_labels.value_counts())

print("\nClass percentages:")
print(
    session_labels.value_counts(normalize=True) * 100
)


# ============================================================
# 10. TRAIN / TEST SPLIT BY CHARGING SESSION
# ============================================================
# IMPORTANT:
# We split entire sessions instead of individual rows.
#
# This prevents data from the same charging session appearing
# in both training and testing data.

train_ids, test_ids = train_test_split(
    session_labels.index,
    test_size=0.20,
    random_state=42,
    stratify=session_labels.values
)

print("\nTraining sessions:", len(train_ids))
print("Testing sessions:", len(test_ids))


# ============================================================
# 11. SCALE FEATURES
# ============================================================
# Fit scaler ONLY on training data.

lstm_scaler = StandardScaler()

train_rows = lstm_df[
    lstm_df["transaction_id"].isin(train_ids)
][time_features]

lstm_scaler.fit(train_rows)

lstm_df[time_features] = (
    lstm_scaler.transform(
        lstm_df[time_features]
    )
)


# ============================================================
# 12. CREATE FIXED-LENGTH TIME SEQUENCES
# ============================================================
# LSTM requires sequences.
#
# We use 60 time points for every charging session.
#
# If a session has fewer than 60 points, the remaining
# positions are zero-padded.

SEQUENCE_LENGTH = 60

X_sequences = []
y_sequences = []
sequence_ids = []

for transaction_id, group in lstm_df.groupby(
    "transaction_id"
):

    values = group[time_features].values

    label = group["label"].iloc[0]

    # --------------------------------------------------------
    # Session has at least 60 time points
    # --------------------------------------------------------

    if len(values) >= SEQUENCE_LENGTH:

        sequence = values[:SEQUENCE_LENGTH]

    # --------------------------------------------------------
    # Session has fewer than 60 points
    # --------------------------------------------------------

    else:

        padding = np.zeros(
            (
                SEQUENCE_LENGTH - len(values),
                len(time_features)
            )
        )

        sequence = np.vstack(
            [values, padding]
        )

    X_sequences.append(sequence)

    y_sequences.append(label)

    sequence_ids.append(transaction_id)


X_sequences = np.array(X_sequences)

y_sequences = np.array(y_sequences)

sequence_ids = np.array(sequence_ids)


print("\nSequence data created:")
print("X shape:", X_sequences.shape)
print("y shape:", y_sequences.shape)


# ============================================================
# 13. SPLIT SEQUENCES INTO TRAIN AND TEST
# ============================================================

train_mask = np.isin(
    sequence_ids,
    train_ids
)

test_mask = np.isin(
    sequence_ids,
    test_ids
)

X_train_lstm = X_sequences[train_mask]

y_train_lstm = y_sequences[train_mask]

X_test_lstm = X_sequences[test_mask]

y_test_lstm = y_sequences[test_mask]


print("\nTraining sequence shape:",
      X_train_lstm.shape)

print("Testing sequence shape:",
      X_test_lstm.shape)


# ============================================================
# 14. BUILD LSTM MODEL
# ============================================================

lstm_model = Sequential([

    LSTM(
        64,
        input_shape=(
            SEQUENCE_LENGTH,
            len(time_features)
        ),
        return_sequences=True
    ),

    Dropout(0.2),

    LSTM(32),

    Dropout(0.2),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])


# ============================================================
# 15. COMPILE MODEL
# ============================================================

lstm_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


print("\n")
print("=" * 60)
print("LSTM MODEL")
print("=" * 60)

lstm_model.summary()


# ============================================================
# 16. TRAIN LSTM
# ============================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)


print("\nStarting LSTM training...")

history = lstm_model.fit(
    X_train_lstm,
    y_train_lstm,

    validation_split=0.15,

    epochs=15,

    batch_size=32,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# ============================================================
# 17. PREDICT ON UNSEEN TEST DATA
# ============================================================

lstm_probability = (
    lstm_model
    .predict(X_test_lstm)
    .ravel()
)

lstm_prediction = (
    lstm_probability >= 0.5
).astype(int)


# ============================================================
# 18. CALCULATE PERFORMANCE METRICS
# ============================================================

lstm_accuracy = accuracy_score(
    y_test_lstm,
    lstm_prediction
)

lstm_precision = precision_score(
    y_test_lstm,
    lstm_prediction,
    zero_division=0
)

lstm_recall = recall_score(
    y_test_lstm,
    lstm_prediction,
    zero_division=0
)

lstm_f1 = f1_score(
    y_test_lstm,
    lstm_prediction,
    zero_division=0
)

lstm_roc_auc = roc_auc_score(
    y_test_lstm,
    lstm_probability
)

# R² is included because your project comparison
# specifically asks for it.
lstm_r2 = r2_score(
    y_test_lstm,
    lstm_probability
)


# ============================================================
# 19. DISPLAY FINAL LSTM RESULTS
# ============================================================

print("\n")
print("=" * 60)
print("LSTM FINAL RESULTS")
print("=" * 60)

print(
    f"Accuracy       : {lstm_accuracy:.4f}"
)

print(
    f"Precision      : {lstm_precision:.4f}"
)

print(
    f"Recall         : {lstm_recall:.4f}"
)

print(
    f"F1 Score       : {lstm_f1:.4f}"
)

print(
    f"ROC-AUC        : {lstm_roc_auc:.4f}"
)

print(
    f"R² Score       : {lstm_r2:.4f}"
)

print("=" * 60)


# ============================================================
# 20. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test_lstm,
        lstm_prediction,
        target_names=[
            "Normal",
            "Faulty"
        ],
        zero_division=0
    )
)


# ============================================================
# 21. CONFUSION MATRIX
# ============================================================

cm_lstm = confusion_matrix(
    y_test_lstm,
    lstm_prediction
)

print("\nConfusion Matrix:")
print(cm_lstm)


# ============================================================
# 22. CONFUSION MATRIX PLOT
# ============================================================

ConfusionMatrixDisplay(
    confusion_matrix=cm_lstm,
    display_labels=[
        "Normal",
        "Faulty"
    ]
).plot()

plt.title(
    "LSTM - EV Charging Fault Detection"
)

plt.show()


# ============================================================
# 23. TRAINING ACCURACY CURVE
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "LSTM Training and Validation Accuracy"
)

plt.legend()

plt.show()


# ============================================================
# 24. TRAINING LOSS CURVE
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "LSTM Training and Validation Loss"
)

plt.legend()

plt.show()


# ============================================================
# 25. CREATE LSTM RESULTS TABLE
# ============================================================

lstm_results = pd.DataFrame({

    "Model": ["LSTM"],

    "Accuracy": [
        lstm_accuracy
    ],

    "Precision": [
        lstm_precision
    ],

    "Recall": [
        lstm_recall
    ],

    "F1 Score": [
        lstm_f1
    ],

    "ROC-AUC": [
        lstm_roc_auc
    ],

    "R2 Score": [
        lstm_r2
    ]
})


print("\n")
print("=" * 60)
print("LSTM RESULTS TABLE")
print("=" * 60)

display(lstm_results)


# ============================================================
# 26. SAVE RESULTS
# ============================================================

RESULTS_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "lstm_results.csv"
)

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "lstm_model.keras"
)

lstm_results.to_csv(
    RESULTS_PATH,
    index=False
)

lstm_model.save(
    MODEL_PATH
)


print("\n")
print("=" * 60)
print("FILES SAVED SUCCESSFULLY")
print("=" * 60)

print("Results:", RESULTS_PATH)

print("Model:", MODEL_PATH)
```
