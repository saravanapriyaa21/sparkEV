```python
# ============================================================
# SPARKEV MODULE 2
# MODEL 3: TEMPORAL CONVOLUTIONAL NETWORK (TCN)
# EV CHARGING FAULT DETECTION
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import drive

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    r2_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

import tensorflow as tf

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Conv1D,
    BatchNormalization,
    Activation,
    Dropout,
    Add,
    GlobalAveragePooling1D,
    Dense
)

from tensorflow.keras.callbacks import EarlyStopping

print("TensorFlow version:", tf.__version__)


# ============================================================
# 2. MOUNT GOOGLE DRIVE
# ============================================================

drive.mount("/content/drive")


# ============================================================
# 3. LOAD DATASET
# ============================================================

DATA_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "processed_data_longer_than_30.xlsx"
)

df = pd.read_excel(DATA_PATH)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Dataset shape:", df.shape)

print(
    "Number of charging sessions:",
    df["transaction_id"].nunique()
)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 4. PREPARE TIME-SERIES DATA
# ============================================================

tcn_df = df.copy()

tcn_df["end_time"] = pd.to_datetime(
    tcn_df["end_time"]
)

tcn_df = tcn_df.sort_values(
    ["transaction_id", "end_time"]
).reset_index(drop=True)


# ============================================================
# 5. FEATURES USED BY TCN
# ============================================================
# These represent EV charging and battery behaviour.

time_features = [
    "current_soc",
    "chargingv",
    "charginga",
    "out_power",
    "charging_gun_temperature1",
    "charging_gun_temperature2",
    "current_energy_meter_value",
    "total_charging_kwh",
    "total_charging_min"
]

print("\nTCN input features:")

for feature in time_features:
    print("-", feature)


# ============================================================
# 6. KEEP REQUIRED COLUMNS
# ============================================================

required_columns = (
    ["transaction_id", "label"] +
    time_features
)

tcn_df = tcn_df[
    required_columns
].copy()


# ============================================================
# 7. HANDLE MISSING VALUES
# ============================================================

print("\nMissing values before cleaning:")

print(
    tcn_df[time_features]
    .isnull()
    .sum()
)

tcn_df = (
    tcn_df
    .dropna()
    .reset_index(drop=True)
)

print(
    "\nShape after removing missing values:",
    tcn_df.shape
)


# ============================================================
# 8. CREATE SESSION-LEVEL LABELS
# ============================================================

session_labels = (
    tcn_df
    .groupby("transaction_id")["label"]
    .first()
)

print("\nSession-level class distribution:")

print(
    session_labels.value_counts()
)

print("\nClass percentages:")

print(
    session_labels
    .value_counts(normalize=True)
    .mul(100)
)


# ============================================================
# 9. TRAIN / TEST SPLIT BY SESSION
# ============================================================
# We split entire charging sessions.
#
# This prevents data from the same charging session from
# appearing in both training and testing sets.

train_ids, test_ids = train_test_split(
    session_labels.index,
    test_size=0.20,
    random_state=42,
    stratify=session_labels.values
)

print("\nTraining sessions:", len(train_ids))

print("Testing sessions:", len(test_ids))


# ============================================================
# 10. STANDARDIZE FEATURES
# ============================================================
# Fit scaler ONLY on training rows.

tcn_scaler = StandardScaler()

train_rows = tcn_df[
    tcn_df["transaction_id"].isin(train_ids)
][time_features]

tcn_scaler.fit(train_rows)

tcn_df[time_features] = (
    tcn_scaler.transform(
        tcn_df[time_features]
    )
)


# ============================================================
# 11. CREATE FIXED-LENGTH SEQUENCES
# ============================================================
# TCN works on sequences.
#
# We use 60 time points for every charging session.
#
# Sessions shorter than 60 points are zero-padded.

SEQUENCE_LENGTH = 60

X_sequences = []

y_sequences = []

sequence_ids = []


for transaction_id, group in tcn_df.groupby(
    "transaction_id"
):

    values = group[
        time_features
    ].values

    label = group[
        "label"
    ].iloc[0]


    # --------------------------------------------------------
    # If session has 60 or more points
    # --------------------------------------------------------

    if len(values) >= SEQUENCE_LENGTH:

        sequence = values[
            :SEQUENCE_LENGTH
        ]


    # --------------------------------------------------------
    # If session has fewer than 60 points
    # --------------------------------------------------------

    else:

        padding = np.zeros(
            (
                SEQUENCE_LENGTH - len(values),
                len(time_features)
            )
        )

        sequence = np.vstack(
            [
                values,
                padding
            ]
        )


    X_sequences.append(sequence)

    y_sequences.append(label)

    sequence_ids.append(transaction_id)


X_sequences = np.array(
    X_sequences
)

y_sequences = np.array(
    y_sequences
)

sequence_ids = np.array(
    sequence_ids
)


print("\n")
print("=" * 60)
print("SEQUENCE DATA")
print("=" * 60)

print(
    "X shape:",
    X_sequences.shape
)

print(
    "y shape:",
    y_sequences.shape
)


# ============================================================
# 12. SPLIT INTO TRAIN AND TEST
# ============================================================

train_mask = np.isin(
    sequence_ids,
    train_ids
)

test_mask = np.isin(
    sequence_ids,
    test_ids
)


X_train_tcn = X_sequences[
    train_mask
]

y_train_tcn = y_sequences[
    train_mask
]

X_test_tcn = X_sequences[
    test_mask
]

y_test_tcn = y_sequences[
    test_mask
]


print("\nTraining sequence shape:")
print(X_train_tcn.shape)

print("\nTesting sequence shape:")
print(X_test_tcn.shape)


# ============================================================
# 13. BUILD TCN RESIDUAL BLOCK
# ============================================================

def tcn_residual_block(
    x,
    filters,
    kernel_size,
    dilation_rate,
    dropout_rate=0.2
):

    # --------------------------------------------------------
    # First causal dilated convolution
    # --------------------------------------------------------

    conv1 = Conv1D(
        filters=filters,
        kernel_size=kernel_size,
        padding="causal",
        dilation_rate=dilation_rate
    )(x)

    conv1 = BatchNormalization()(conv1)

    conv1 = Activation(
        "relu"
    )(conv1)

    conv1 = Dropout(
        dropout_rate
    )(conv1)


    # --------------------------------------------------------
    # Second causal dilated convolution
    # --------------------------------------------------------

    conv2 = Conv1D(
        filters=filters,
        kernel_size=kernel_size,
        padding="causal",
        dilation_rate=dilation_rate
    )(conv1)

    conv2 = BatchNormalization()(conv2)

    conv2 = Activation(
        "relu"
    )(conv2)

    conv2 = Dropout(
        dropout_rate
    )(conv2)


    # --------------------------------------------------------
    # Residual connection
    # --------------------------------------------------------

    if x.shape[-1] != filters:

        residual = Conv1D(
            filters=filters,
            kernel_size=1,
            padding="same"
        )(x)

    else:

        residual = x


    output = Add()(
        [
            residual,
            conv2
        ]
    )

    output = Activation(
        "relu"
    )(output)


    return output


# ============================================================
# 14. BUILD COMPLETE TCN MODEL
# ============================================================

input_layer = Input(
    shape=(
        SEQUENCE_LENGTH,
        len(time_features)
    )
)


# First temporal block
x = tcn_residual_block(
    input_layer,
    filters=32,
    kernel_size=3,
    dilation_rate=1,
    dropout_rate=0.2
)


# Second temporal block
x = tcn_residual_block(
    x,
    filters=32,
    kernel_size=3,
    dilation_rate=2,
    dropout_rate=0.2
)


# Third temporal block
x = tcn_residual_block(
    x,
    filters=32,
    kernel_size=3,
    dilation_rate=4,
    dropout_rate=0.2
)


# Fourth temporal block
x = tcn_residual_block(
    x,
    filters=32,
    kernel_size=3,
    dilation_rate=8,
    dropout_rate=0.2
)


# ============================================================
# 15. GLOBAL TEMPORAL REPRESENTATION
# ============================================================

x = GlobalAveragePooling1D()(x)


# ============================================================
# 16. FULLY CONNECTED CLASSIFICATION LAYER
# ============================================================

x = Dense(
    32,
    activation="relu"
)(x)

x = Dropout(
    0.2
)(x)


# ============================================================
# 17. OUTPUT LAYER
# ============================================================

output_layer = Dense(
    1,
    activation="sigmoid"
)(x)


# ============================================================
# 18. CREATE MODEL
# ============================================================

tcn_model = Model(
    inputs=input_layer,
    outputs=output_layer
)


# ============================================================
# 19. COMPILE MODEL
# ============================================================

tcn_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


print("\n")
print("=" * 60)
print("TCN MODEL")
print("=" * 60)

tcn_model.summary()


# ============================================================
# 20. EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)


# ============================================================
# 21. TRAIN TCN
# ============================================================

print("\nStarting TCN training...")

history = tcn_model.fit(

    X_train_tcn,

    y_train_tcn,

    validation_split=0.15,

    epochs=15,

    batch_size=32,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# ============================================================
# 22. PREDICT ON UNSEEN TEST DATA
# ============================================================

tcn_probability = (
    tcn_model
    .predict(
        X_test_tcn
    )
    .ravel()
)


tcn_prediction = (
    tcn_probability >= 0.5
).astype(int)


# ============================================================
# 23. CALCULATE PERFORMANCE METRICS
# ============================================================

tcn_accuracy = accuracy_score(
    y_test_tcn,
    tcn_prediction
)

tcn_precision = precision_score(
    y_test_tcn,
    tcn_prediction,
    zero_division=0
)

tcn_recall = recall_score(
    y_test_tcn,
    tcn_prediction,
    zero_division=0
)

tcn_f1 = f1_score(
    y_test_tcn,
    tcn_prediction,
    zero_division=0
)

tcn_roc_auc = roc_auc_score(
    y_test_tcn,
    tcn_probability
)

# R² included because it was requested
# for project comparison.

tcn_r2 = r2_score(
    y_test_tcn,
    tcn_probability
)


# ============================================================
# 24. DISPLAY FINAL RESULTS
# ============================================================

print("\n")
print("=" * 60)
print("TCN FINAL RESULTS")
print("=" * 60)

print(
    f"Accuracy       : {tcn_accuracy:.4f}"
)

print(
    f"Precision      : {tcn_precision:.4f}"
)

print(
    f"Recall         : {tcn_recall:.4f}"
)

print(
    f"F1 Score       : {tcn_f1:.4f}"
)

print(
    f"ROC-AUC        : {tcn_roc_auc:.4f}"
)

print(
    f"R² Score       : {tcn_r2:.4f}"
)

print("=" * 60)


# ============================================================
# 25. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test_tcn,
        tcn_prediction,
        target_names=[
            "Normal",
            "Faulty"
        ],
        zero_division=0
    )
)


# ============================================================
# 26. CONFUSION MATRIX
# ============================================================

cm_tcn = confusion_matrix(
    y_test_tcn,
    tcn_prediction
)

print("\nConfusion Matrix:")

print(cm_tcn)


# ============================================================
# 27. CONFUSION MATRIX PLOT
# ============================================================

ConfusionMatrixDisplay(
    confusion_matrix=cm_tcn,
    display_labels=[
        "Normal",
        "Faulty"
    ]
).plot()

plt.title(
    "TCN - EV Charging Fault Detection"
)

plt.show()


# ============================================================
# 28. TRAINING ACCURACY CURVE
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "TCN Training and Validation Accuracy"
)

plt.legend()

plt.show()


# ============================================================
# 29. TRAINING LOSS CURVE
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "TCN Training and Validation Loss"
)

plt.legend()

plt.show()


# ============================================================
# 30. CREATE RESULTS TABLE
# ============================================================

tcn_results = pd.DataFrame({

    "Model": [
        "TCN"
    ],

    "Accuracy": [
        tcn_accuracy
    ],

    "Precision": [
        tcn_precision
    ],

    "Recall": [
        tcn_recall
    ],

    "F1 Score": [
        tcn_f1
    ],

    "ROC-AUC": [
        tcn_roc_auc
    ],

    "R2 Score": [
        tcn_r2
    ]
})


print("\n")
print("=" * 60)
print("TCN RESULTS TABLE")
print("=" * 60)

display(
    tcn_results
)


# ============================================================
# 31. SAVE RESULTS
# ============================================================

RESULTS_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "tcn_results.csv"
)

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "tcn_model.keras"
)

SCALER_PATH = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "tcn_scaler.pkl"
)


tcn_results.to_csv(
    RESULTS_PATH,
    index=False
)


tcn_model.save(
    MODEL_PATH
)


# Save scaler
import joblib

joblib.dump(
    tcn_scaler,
    SCALER_PATH
)


print("\n")
print("=" * 60)
print("TCN FILES SAVED SUCCESSFULLY")
print("=" * 60)

print(
    "Results:",
    RESULTS_PATH
)

print(
    "Model:",
    MODEL_PATH
)

print(
    "Scaler:",
    SCALER_PATH
)

print("=" * 60)
```
