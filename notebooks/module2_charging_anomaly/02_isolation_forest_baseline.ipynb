```python
session_features = df.groupby("transaction_id").agg(

    # Target
    label=("label", "first"),

    # Battery type
    battery_type=("types", "first"),

    # SOC features
    soc_start=("current_soc", "first"),
    soc_end=("current_soc", "last"),
    soc_min=("current_soc", "min"),
    soc_max=("current_soc", "max"),
    soc_mean=("current_soc", "mean"),
    soc_std=("current_soc", "std"),

    # Voltage
    voltage_start=("chargingv", "first"),
    voltage_end=("chargingv", "last"),
    voltage_min=("chargingv", "min"),
    voltage_max=("chargingv", "max"),
    voltage_mean=("chargingv", "mean"),
    voltage_std=("chargingv", "std"),

    # Current
    current_start=("charginga", "first"),
    current_end=("charginga", "last"),
    current_min=("charginga", "min"),
    current_max=("charginga", "max"),
    current_mean=("charginga", "mean"),
    current_std=("charginga", "std"),

    # Power
    power_start=("out_power", "first"),
    power_end=("out_power", "last"),
    power_min=("out_power", "min"),
    power_max=("out_power", "max"),
    power_mean=("out_power", "mean"),
    power_std=("out_power", "std"),

    # Temperature sensor 1
    temp1_start=("charging_gun_temperature1", "first"),
    temp1_end=("charging_gun_temperature1", "last"),
    temp1_min=("charging_gun_temperature1", "min"),
    temp1_max=("charging_gun_temperature1", "max"),
    temp1_mean=("charging_gun_temperature1", "mean"),
    temp1_std=("charging_gun_temperature1", "std"),

    # Temperature sensor 2
    temp2_start=("charging_gun_temperature2", "first"),
    temp2_end=("charging_gun_temperature2", "last"),
    temp2_min=("charging_gun_temperature2", "min"),
    temp2_max=("charging_gun_temperature2", "max"),
    temp2_mean=("charging_gun_temperature2", "mean"),
    temp2_std=("charging_gun_temperature2", "std"),

    # Energy
    energy_start=("total_charging_kwh", "first"),
    energy_end=("total_charging_kwh", "last"),
    energy_max=("total_charging_kwh", "max"),

    # Charging duration
    duration_min=("total_charging_min", "min"),
    duration_max=("total_charging_min", "max"),

    # Number of observations
    num_points=("id", "count")
)

print("Session-level dataset created!")
print("Shape:", session_features.shape)
```

```python
import numpy as np
# SOC change
session_features["soc_gain"] = (
    session_features["soc_end"] -
    session_features["soc_start"]
)

# Voltage change
session_features["voltage_change"] = (
    session_features["voltage_end"] -
    session_features["voltage_start"]
)

# Current change
session_features["current_change"] = (
    session_features["current_end"] -
    session_features["current_start"]
)

# Power change
session_features["power_change"] = (
    session_features["power_end"] -
    session_features["power_start"]
)

# Temperature rise
session_features["temp1_rise"] = (
    session_features["temp1_max"] -
    session_features["temp1_start"]
)

session_features["temp2_rise"] = (
    session_features["temp2_max"] -
    session_features["temp2_start"]
)

# Temperature sensor difference
session_features["temp_sensor_difference"] = (
    session_features["temp1_mean"] -
    session_features["temp2_mean"]
)

# Energy delivered during session
session_features["energy_delivered"] = (
    session_features["energy_end"] -
    session_features["energy_start"]
)

# Energy per SOC gained
session_features["energy_per_soc"] = (
    session_features["energy_delivered"] /
    session_features["soc_gain"].replace(0, np.nan)
)

print("Derived features created!")
```

```python
# Create a copy for modelling
model_data = session_features.copy()

# Separate the true labels for later evaluation
y = model_data["label"].copy()

# Remove target and non-predictive identifier
X = model_data.drop(
    columns=["label"],
    errors="ignore"
)

print("Features before cleanup:", X.shape)
print("Target shape:", y.shape)
```

```python
drop_columns = [
    "battery_type",
    "duration_min"
]

X_numeric = X.drop(columns=drop_columns, errors="ignore")

print("Numeric features:", X_numeric.shape)
```

```python
battery_encoded = pd.get_dummies(
    session_features["battery_type"],
    prefix="battery_type",
    dtype=int
)

print(battery_encoded.head())
```

```python
X_model = pd.concat(
    [
        X_numeric.reset_index(drop=True),
        battery_encoded.reset_index(drop=True)
    ],
    axis=1
)

print("Final modelling dataset shape:", X_model.shape)
```

```python
X_model = X_model.replace(
    [np.inf, -np.inf],
    np.nan
)

X_model = X_model.fillna(
    X_model.median(numeric_only=True)
)

print("Remaining missing values:",
      X_model.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

normal_indices = y[y == 0].index
faulty_indices = y[y == 1].index

normal_train_idx, normal_test_idx = train_test_split(
    normal_indices,
    test_size=0.30,
    random_state=42
)

print("Normal training sessions:", len(normal_train_idx))
print("Normal test sessions:", len(normal_test_idx))
print("Faulty sessions available for evaluation:", len(faulty_indices))
```

```python
# Reset the index so X and y use the same 0,1,2,... index

X_model = X_model.reset_index(drop=True)
y = y.reset_index(drop=True)

print("X_model shape:", X_model.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_model,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining label distribution:")
print(y_train.value_counts())

print("\nTesting label distribution:")
print(y_test.value_counts())
```

```python
# Keep only normal sessions for unsupervised training

X_train_normal = X_train[y_train == 0]

print("Normal sessions used for model training:",
      X_train_normal.shape[0])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_normal_scaled = scaler.fit_transform(X_train_normal)
X_test_scaled = scaler.transform(X_test)

print("Training data:", X_train_normal_scaled.shape)
print("Test data:", X_test_scaled.shape)
```

```python
from sklearn.ensemble import IsolationForest

iso_forest = IsolationForest(
    n_estimators=300,
    max_samples="auto",
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

iso_forest.fit(X_train_normal_scaled)

print("Isolation Forest trained successfully!")
```

```python
normality_score = iso_forest.decision_function(X_test_scaled)

# Higher value = more anomalous
anomaly_score = -normality_score

results = pd.DataFrame({
    "actual_label": y_test.values,
    "anomaly_score": anomaly_score
})

results.head()
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

# Use the model's built-in predictions
# -1 = anomaly
#  1 = normal
iso_predictions = iso_forest.predict(X_test_scaled)

# Convert to our label format
# anomaly = 1, normal = 0
y_pred = (iso_predictions == -1).astype(int)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Normal", "Faulty"]
))
```

```python
roc_auc = roc_auc_score(y_test, anomaly_score)
pr_auc = average_precision_score(y_test, anomaly_score)

print("ROC-AUC:", round(roc_auc, 4))
print("PR-AUC:", round(pr_auc, 4))
```

```python
score_comparison = results.groupby("actual_label")["anomaly_score"].describe()

print(score_comparison)
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=results,
    x="actual_label",
    y="anomaly_score"
)

plt.xlabel("Actual Label (0 = Normal, 1 = Faulty)")
plt.ylabel("Anomaly Score")
plt.title("Isolation Forest Anomaly Scores")
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_test,
    anomaly_score
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_index = np.argmax(f1_scores)

best_threshold = thresholds[best_index]
best_f1 = f1_scores[best_index]

print("Best anomaly threshold:", best_threshold)
print("Best F1-score:", best_f1)
```

```python
y_pred_tuned = (
    anomaly_score >= best_threshold
).astype(int)

print(classification_report(
    y_test,
    y_pred_tuned,
    target_names=["Normal", "Faulty"]
))
```

```python
from sklearn.model_selection import train_test_split

# Split the original training set into:
# 90% model training and 10% validation

X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.10,
    stratify=y_train,
    random_state=42
)

print("Model training samples:", X_fit.shape[0])
print("Validation samples:", X_val.shape[0])

print("\nValidation label distribution:")
print(y_val.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

final_scaler = StandardScaler()

X_fit_normal_scaled = final_scaler.fit_transform(X_fit_normal)

X_val_scaled = final_scaler.transform(X_val)
X_test_final_scaled = final_scaler.transform(X_test)

print("Training:", X_fit_normal_scaled.shape)
print("Validation:", X_val_scaled.shape)
print("Test:", X_test_final_scaled.shape)
```

```python
from sklearn.ensemble import IsolationForest

final_iso_forest = IsolationForest(
    n_estimators=300,
    max_samples="auto",
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

final_iso_forest.fit(X_fit_normal_scaled)

print("Final Isolation Forest trained successfully!")
```

```python
val_normality_score = final_iso_forest.decision_function(
    X_val_scaled
)

val_anomaly_score = -val_normality_score

print("Validation anomaly scores generated.")
```

```python
from sklearn.metrics import precision_recall_curve
import numpy as np

precision_val, recall_val, thresholds_val = precision_recall_curve(
    y_val,
    val_anomaly_score
)

f1_val = (
    2 * precision_val[:-1] * recall_val[:-1]
    / (precision_val[:-1] + recall_val[:-1] + 1e-10)
)

best_val_index = np.argmax(f1_val)

final_threshold = thresholds_val[best_val_index]

print("Selected threshold:", final_threshold)
print("Validation F1-score:", f1_val[best_val_index])
```

```python
# Generate anomaly scores for the untouched test set

test_normality_score = final_iso_forest.decision_function(
    X_test_final_scaled
)

test_anomaly_score = -test_normality_score

# Apply threshold selected from validation data

y_test_final_pred = (
    test_anomaly_score >= final_threshold
).astype(int)

print("Final Test Classification Report:")

print(
    classification_report(
        y_test,
        y_test_final_pred,
        target_names=["Normal", "Faulty"]
    )
)
```

```python
from sklearn.metrics import (
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_final_pred))

final_roc_auc = roc_auc_score(
    y_test,
    test_anomaly_score
)

final_pr_auc = average_precision_score(
    y_test,
    test_anomaly_score
)

print("\nFinal ROC-AUC:", round(final_roc_auc, 4))
print("Final PR-AUC:", round(final_pr_auc, 4))
```

```python
# ============================================================
# EXPORT COMPLETE ISOLATION FOREST MODEL FOR ESP32
# ============================================================

import json
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# ------------------------------------------------------------
# 1. MAKE SURE THESE ARE THE EXACT OBJECTS FROM YOUR FINAL MODEL
# ------------------------------------------------------------

model = final_iso_forest
model_scaler = scaler

# IMPORTANT:
# Replace this ONLY if your feature list has a different variable name.
feature_names = list(X_model.columns)

# ------------------------------------------------------------
# 2. GET THE FINAL CLASSIFICATION THRESHOLD
# ------------------------------------------------------------
# This threshold must be the one selected using the VALIDATION set,
# NOT the test set.

# If you already created a validation threshold variable, use it.
# Example variable names that may exist in your notebook:
#
# validation_threshold
# best_threshold
# optimal_threshold
#
# The code below tries to find one automatically.

possible_threshold_names = [
    "validation_threshold",
    "best_threshold",
    "optimal_threshold",
    "selected_threshold",
    "decision_threshold"
]

decision_threshold = None

for name in possible_threshold_names:
    if name in globals():
        decision_threshold = float(globals()[name])
        print(f"Using validation threshold: {name} = {decision_threshold}")
        break

if decision_threshold is None:
    raise ValueError(
        "No validation threshold was found. "
        "Set decision_threshold manually to the threshold selected "
        "from the validation set before exporting the JSON."
    )

# ------------------------------------------------------------
# 3. BASIC MODEL INFORMATION
# ------------------------------------------------------------

export_data = {
    "model_type": "IsolationForest",
    "model_version": "SPARKEV_Module2_Final",
    "feature_count": len(feature_names),
    "feature_names": feature_names,

    # --------------------------------------------------------
    # 4. PREPROCESSING INFORMATION
    # --------------------------------------------------------

    "preprocessing": {
        "scaler_type": "StandardScaler",
        "mean": model_scaler.mean_.tolist(),
        "scale": model_scaler.scale_.tolist()
    },

    # --------------------------------------------------------
    # 5. ISOLATION FOREST PARAMETERS
    # --------------------------------------------------------

    "model_parameters": {
        "n_estimators": int(model.n_estimators),
        "max_samples": (
            int(model.max_samples_)
            if hasattr(model, "max_samples_")
            else str(model.max_samples)
        ),
        "contamination": (
            float(model.contamination)
            if isinstance(model.contamination, (int, float))
            else str(model.contamination)
        ),
        "random_state": (
            int(model.random_state)
            if model.random_state is not None
            else None
        )
    },

    # --------------------------------------------------------
    # 6. FINAL DECISION THRESHOLD
    # --------------------------------------------------------
    #
    # This is DIFFERENT from the thresholds inside each tree.
    #
    # It is the threshold used after calculating the overall
    # Isolation Forest anomaly score.
    #

    "decision": {
        "threshold": decision_threshold,
        "threshold_source": "validation_set",
        "score_rule": (
            "Anomaly if anomaly_score >= threshold; "
            "Normal otherwise"
        ),
        "predicted_label_rule": {
            "0": "Normal",
            "1": "Faulty / Anomalous"
        }
    },

    # --------------------------------------------------------
    # 7. ISOLATION FOREST OFFSET
    # --------------------------------------------------------

    "isolation_forest": {
        "offset": float(model.offset_),
        "estimators_count": len(model.estimators_),
        "trees": []
    }
}

# ------------------------------------------------------------
# 8. EXPORT EVERY TREE
# ------------------------------------------------------------

print("Exporting trees...")

for tree_index, estimator in enumerate(model.estimators_):

    tree = estimator.tree_

    tree_data = {
        "tree_index": int(tree_index),
        "node_count": int(tree.node_count),

        "children_left": tree.children_left.astype(int).tolist(),
        "children_right": tree.children_right.astype(int).tolist(),

        # Feature index used at every split
        "feature": tree.feature.astype(int).tolist(),

        # Threshold used at every split
        "threshold": tree.threshold.astype(float).tolist(),

        # Number of training samples reaching every node
        "n_node_samples": tree.n_node_samples.astype(int).tolist()
    }

    export_data["isolation_forest"]["trees"].append(tree_data)

# ------------------------------------------------------------
# 9. ADD FEATURE INDEX MAP
# ------------------------------------------------------------
# Useful for ESP32 so the firmware knows exactly which index
# corresponds to which feature.

export_data["feature_index_map"] = {
    str(index): name
    for index, name in enumerate(feature_names)
}

# ------------------------------------------------------------
# 10. ADD EXPECTED INPUT FORMAT
# ------------------------------------------------------------

export_data["input_format"] = {
    "description": (
        "Input must contain the 57 session-level features "
        "in exactly the order specified by feature_names."
    ),
    "feature_count": len(feature_names),
    "requires_standard_scaling": True,
    "scaling_formula": (
        "scaled_value = (raw_value - mean) / scale"
    )
}

# ------------------------------------------------------------
# 11. ADD OUTPUT FORMAT
# ------------------------------------------------------------

export_data["output_format"] = {
    "anomaly_score": (
        "Isolation Forest anomaly score calculated from "
        "the ensemble of trees"
    ),
    "predicted_label": {
        "0": "Normal",
        "1": "Faulty / Anomalous"
    },
    "classification": (
        "predicted_label = 1 when anomaly_score >= decision threshold"
    )
}

# ------------------------------------------------------------
# 12. ADD TRAINING INFORMATION
# ------------------------------------------------------------

export_data["training_information"] = {
    "training_method": "Unsupervised anomaly detection",
    "training_data": "Normal charging sessions only",
    "session_level_model": True,
    "data_granularity": "Charging session",
    "project": "SPARKEV Module 2",
    "purpose": (
        "Detect abnormal EV charging behaviour using "
        "multiple charging and battery-related factors."
    )
}

# ------------------------------------------------------------
# 13. SAVE JSON
# ------------------------------------------------------------

json_path = (
    "/content/drive/MyDrive/"
    "Anomoly_dataset/"
    "isolation_forest_esp32_complete.json"
)

with open(json_path, "w") as f:
    json.dump(export_data, f, indent=4)

print("\n================================================")
print("JSON EXPORT COMPLETED SUCCESSFULLY")
print("================================================")
print(f"File: {json_path}")
print(f"Features: {len(feature_names)}")
print(f"Trees: {len(model.estimators_)}")
print(f"Decision threshold: {decision_threshold}")
print(f"Offset: {model.offset_}")
print("================================================")
```
